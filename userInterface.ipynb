Import packages

```python
import pandas as pd
from ipyleaflet import Map, Marker
from ipywidgets import Output, Label, VBox, HTML, Layout
```

Load in .csv file

```python
trafficStations = pd.read_csv("datasets_cleaned/trafficStations.csv")
```

Remove unnessessary columns

```python
trafficStations = trafficStations[['station_key', 'full_name', 'wgs84_latitude', 'wgs84_longitude']]
```

Create an interactive map showing all station locations with names as popups. When a pin is clicked, a function runs that returns the station key.

```python
# Dictionary to store the selected station key
selected = {'key': None}

# Label widget to display the selected station key
label = Label("Selected station key: None")

# Create a map centered at a default location with increased height
m = Map(
    center=(-34, 151),
    zoom=10,
    layout=Layout(height="1200px")  # Set height using layout for better effect
)

# Function to handle marker click events
def make_on_click(station_key):
    def on_click(event=None, **kwargs):
        selected['key'] = station_key
        label.value = f"Selected station key: {selected['key']}"
    return on_click

# Add a marker for each station
for idx, row in trafficStations.iterrows():
    marker = Marker(
        location=(row['wgs84_latitude'], row['wgs84_longitude']),
        title=row['full_name'],
        draggable=False  # Prevent marker from being moved
    )
    marker.popup = HTML(value=row['full_name'])
    marker.on_click(make_on_click(row['station_key']))
    m.add_layer(marker)

# Display the map and label in a vertical box layout
VBox([m, label])
```
